```python
import json
with open("../data/refined_tables/第八至十编-表格化结果.json", "r", encoding="utf-8") as f:
  data = json.load(f)

text_length = 0
label_count = 0
for entry in data:
  for key, value in entry.items():
    label_count += 1
    text_length += len(value)

print(label_count, text_length)
```

```python
import re

pattern = re.compile(r"词条\s+(.+?)\s+处理完成，用时\s+([\d.]+)\s+秒")

records = []
with open("./nohup.20260304.log", "r", encoding="utf-8", errors="replace") as f:
    for line in f:
        m = pattern.search(line)
        if m:
            records.append(
                {"词条名": m.group(1).strip(), "用时秒": float(m.group(2))}
            )

print(f"共匹配 {len(records)} 条")
if records:
    print("示例（前 5 条）:", records[:5])
    avg_sec = sum(r["用时秒"] for r in records) / len(records)
    print(f"平均用时: {avg_sec:.4f} 秒")
```
